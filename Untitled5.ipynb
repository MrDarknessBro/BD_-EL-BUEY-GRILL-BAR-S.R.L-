```python
# Configuración visual para que los gráficos luzcan profesionales
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
# Iniciar Sesión de Spark
spark = SparkSession.builder.appName("OmniStore_Viz").getOrCreate()
print("¡Entorno listo para graficar!")
```

```python
# Conteo de filas por tabla (BD Buey Grill Bar)
conteo = pd.DataFrame({
    "Tabla": ["DetallePedido","Pedido","Venta","DetalleCompra","EmpleadoTurno","Reserva","Cliente",
              "DireccionCliente","Compra","ProductoInsumo","Producto","Insumo","Mesa","Empleado",
              "Categoria","Proveedor","MetodoPago","Rol","Turno","Zona"],
    "Filas": [71855,23431,22349,4001,3040,1824,1800,1716,800,189,92,22,20,20,13,6,5,4,3,2]
})
```

```python
# Muestra de ventas (top 15) con el método de pago
metodos = {1:"Efectivo", 2:"Tarjeta de Crédito", 3:"Tarjeta de Débito", 4:"Yape", 5:"Plin"}
ventas = pd.DataFrame({
    "IdVenta": range(1,16),
    "IdMetodoPago": [4,4,4,4,4,5,5,5,5,1,1,1,1,3,3],
    "Total": [186.44,247.80,99.12,156.94,87.32,224.79,103.84,154.58,205.32,
              97.94,93.22,173.46,109.74,110.92,38.94]
})
ventas["Metodo"] = ventas["IdMetodoPago"].map(metodos)
```

# **GRÁFICO 1:**

```python
# Gráfico 1: registros por tabla (escala logarítmica por la gran diferencia de tamaños)
datos = conteo.sort_values("Filas")
fig, ax = plt.subplots(figsize=(10, 7))
barras = ax.barh(datos["Tabla"], datos["Filas"], color="#c0392b")
ax.set_xscale("log")
ax.set_title("Cantidad de registros por tabla - BD Buey Grill Bar", fontsize=14, weight="bold")
ax.set_xlabel("Número de filas (escala logarítmica)")
ax.bar_label(barras, labels=[f"{v:,}" for v in datos["Filas"]], padding=3, fontsize=9)
ax.set_xlim(1, 300000)
ax.grid(axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

# **GRÁFICO 2:**

```python
# Gráfico 2: total vendido por método de pago (muestra de 15 ventas)
resumen = ventas.groupby("Metodo")["Total"].sum().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(8, 5))
barras = ax.bar(resumen.index, resumen.values, color=["#2980b9","#27ae60","#f39c12","#8e44ad"])
ax.set_title("Total vendido por método de pago (muestra de 15 ventas)", fontsize=13, weight="bold")
ax.set_ylabel("Total (S/)")
ax.bar_label(barras, labels=[f"S/ {v:,.2f}" for v in resumen.values], padding=3)
ax.set_ylim(0, resumen.max() * 1.15)
ax.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

# **GRÁFICO 3:**

```python
# Top 10 insumos comprados - 2026 (filtro: UnidadMedida = kg)
# Valores aproximados leídos del gráfico del dashboard
insumos = pd.DataFrame({
    "Insumo": ["Papa blanca","Carne molida artesanal","Pechuga de pollo fresca","Hielo","Tomate",
               "Pierna de cerdo","Lechuga","Bife de res","Tira de cerdo","Pierna de pollo deshuesada"],
    "Kg": [343, 319, 230, 156, 121, 110, 90, 78, 60, 47]
}).sort_values("Kg")

fig, ax = plt.subplots(figsize=(10, 6))
barras = ax.barh(insumos["Insumo"], insumos["Kg"], color="#bf8b41")
ax.set_title("Top 10 insumos comprados - 2026 (kg)", fontsize=14, weight="bold")
ax.set_xlabel("Cantidad comprada (kg)")
ax.bar_label(barras, labels=[f"{v} kg" for v in insumos["Kg"]], padding=3, fontsize=9)
ax.set_xlim(0, 400)
ax.grid(axis="x", linestyle="--", alpha=0.4)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
# **GRÁFICO 4:**

# Ventas trimestrales por producto - 2026 (S/)
ventas_trimestrales = pd.DataFrame({
    "Producto": ["1/4 de Pollo al Cilindro", "16 Piezas con papa frita", "16 Piezas con papa frita y 4 ensaladas",
                 "5 Piezas con papa frita", "5 Piezas con papa frita y 1 ensalada", "8 Piezas con papa frita",
                 "8 Piezas con papa frita y 2 ensaladas", "Adicional 1 chorizo",
                 "Adicional 120gr. cerdo al cilindro", "Adicional 120gr. pechuga a la parrilla"],
    "T1": [550, 1127, 880, 570, 651, 918, 780, 18, 60, 80],
    "T2": [726, 1225, 1155, 722, 945, 891, 960, 60, 70, 150],
    "T3": [594, 1127, 1485, 893, 1071, 1026, 1260, 78, 100, 80],
    "T4": [44, 147, 110, 171, 147, 216, 180, 24, 10, 20],
})
ventas_trimestrales["Total"] = ventas_trimestrales[["T1", "T2", "T3", "T4"]].sum(axis=1)
ventas_trimestrales = ventas_trimestrales.sort_values("Total")

colores = {"T1": "#1e88e5", "T2": "#1a237e", "T3": "#e8743b", "T4": "#6a1b9a"}

fig, ax = plt.subplots(figsize=(11, 6.5))
izquierda = [0] * len(ventas_trimestrales)
for t in ["T1", "T2", "T3", "T4"]:
    ax.barh(ventas_trimestrales["Producto"], ventas_trimestrales[t], left=izquierda, label=f"Trimestre {t[1]}", color=colores[t])
    izquierda = [a + b for a, b in zip(izquierda, ventas_trimestrales[t])]

for i, total in enumerate(ventas_trimestrales["Total"]):
    ax.text(total + 40, i, f"S/ {total:,}", va="center", fontsize=9)

ax.set_title("Ventas trimestrales por producto - 2026 (S/)", fontsize=14, weight="bold")
ax.set_xlabel("Monto vendido (S/)")
ax.set_xlim(0, ventas_trimestrales["Total"].max() * 1.13)
ax.legend(title="Periodo", loc="lower right")
ax.grid(axis="x", linestyle="--", alpha=0.4)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()
```
